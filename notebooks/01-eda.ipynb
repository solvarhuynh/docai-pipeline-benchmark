# 01 - Khảo sát và khám phá dữ liệu (Exploratory Data Analysis - EDA)

Giai đoạn 1: Khảo sát và chuẩn bị dữ liệu thật.

Mục tiêu:
- Khảo sát 4 bộ dữ liệu: mcocr2021, CORD, SROIE, CUAD bằng các hàm chuẩn hoá từ package `docai.data`.
- Đánh giá số lượng mẫu, độ phân giải ảnh, định dạng nhãn, chất lượng ảnh (mờ, nghiêng).
- Chuẩn bị ánh xạ sang JSON schema thống nhất (`docai.core.schema`).

```python
import pandas as pd
from docai.core.config import settings
from docai.data.loaders import list_available_datasets, get_dataset_dir, scan_raw_samples
from docai.data.eda import analyze_dataset, summarize_dataset_structure
from docai.core.schema import DocumentType

print(f"Thư mục dữ liệu thô: {settings.raw_data_dir}")
datasets_info = list_available_datasets()
pd.DataFrame.from_dict(datasets_info, orient="index")[["display_name", "document_type", "is_downloaded", "file_count"]]
```

```python
mcocr_summary = analyze_dataset("mcocr2021", max_samples=20)
print(f"mcocr2021 status: {mcocr_summary.get('status')}")
print(f"Total files: {mcocr_summary.get('total_files')}")
# TODO: Giai đoạn 1 - Tải dữ liệu thật về data/raw/mcocr2021 và chạy phân tích chi tiết
```
